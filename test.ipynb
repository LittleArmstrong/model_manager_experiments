```python
from ultralytics import YOLO
import torch.nn as nn

# Lade Modell mit ALLEN Gewichten (691/691)
model = YOLO('YOLOv12n.pt')

# Füge Dropout hinzu
dropout = nn.Dropout(0.15)
model.model.model[10].register_forward_hook(lambda m, i, o: dropout(o))

# Fertig! Training starten
results = model.train(
    data="datasets/strawberry_test/data.yaml",
    epochs=1,
    freeze=9,
    batch=16
)
```
